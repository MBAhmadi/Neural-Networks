```python
mnist_hist = np.zeros(256, dtype=np.int32)
```

```python
for img in tqdm.tqdm(images_np_list):
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            intensity = img[i,j]

            mnist_hist[intensity] += 1
```

```python
plt.plot(mnist_hist)
plt.show
```
