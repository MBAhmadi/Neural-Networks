```python
def convolve(image, kernel):
    output = np.zeros_like(image, dtype=np.float32)
    padded_image = np.pad(image, pad_width=kernel.shape[0]//2)

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            region = padded_image[i:i+kernel.shape[0], j:j+kernel.shape[1]]
            output[i,j] = np.sum(region*kernel)
    return np.clip(output, 0, 255).astype(np.uint8)
```

```python
def avg_kernel(kernel_size=3):
  kernel = np.ones((kernel_size, kernel_size), dtype=np.float32)
  kernel = kernel/(kernel_size**2)

  return kernel
```

```python
kernel_3 = avg_kernel(3)
kernel_5 = avg_kernel(5)
kernel_7 = avg_kernel(7)
```

```python
filtered_3 = convolve(cameraman, kernel_3)
filtered_5 = convolve(cameraman, kernel_5)
filtered_7 = convolve(cameraman, kernel_7)
```

```python
avg_kernel = np.array([[1/9, 1/9, 1/9],[1/9, 1/9, 1/9],[1/9, 1/9, 1/9]])
```

```python
filtered_image = convolve(cameraman, avg_kernel)
```

```python
plt.imshow(filtered_image, cmap= "gray")
```

```python
plt.figure(figsize=(20,5))
plt.subplot(1,4,1)
plt.imshow(cameraman, cmap="gray")
plt.title("original image")

plt.subplot(1,4,2)
plt.imshow(filtered_3, cmap="gray")
plt.title("kernel size 3")

plt.subplot(1,4,3)
plt.imshow(filtered_5, cmap="gray")
plt.title("kernel size 5")

plt.subplot(1,4,4)
plt.imshow(filtered_7, cmap="gray")
plt.title("kernel size 7")
```
